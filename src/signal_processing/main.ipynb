Frequency Domain Analysis

We used fake signales!

```python
dt = 0.01
t = np.arange(0, 10, dt)
x = 5 + 2 * np.sin(2 * np.pi * 0.5 * t) + 1 * np.sin(2 * np.pi * 4 * t + np.pi / 5)  # frequency = 0.5 Hz, amplitude = 2, offset = 5 , frequency = 4 Hz, amplitude = 1 , phase = pi/5

plt.plot(t, x)
plt.plot(t, np.mean(x) * np.ones_like(t), "r--")  # Baseline (mean) line
plt.title("Synthetic Signal")
plt.xlabel("Time (Sec)")
plt.ylabel("Amplitude")
plt.show()
```

```python
detrend_x = x - np.mean(x)

plt.plot(t, detrend_x)
plt.plot(t, np.mean(detrend_x) * np.ones_like(t), "r--")  # Baseline (mean) line ---> Zero-mean signal
plt.title("Synthetic Signal")
plt.xlabel("Time (Sec)")
plt.ylabel("Amplitude")
plt.show()
```

```python
FFT_x = np.fft.fft(x)
L = len(x)
freq = np.fft.fftfreq(len(x), dt)

# We devided the amplitude by L to normalize the amplitude of the FFT output. This is because the FFT output is scaled by the length of the input signal, and dividing by L ensures that the amplitude of the frequency components is correctly represented in the frequency domain.

plt.subplot(2, 1, 1)
plt.plot(freq[freq >= 0], np.abs(FFT_x.real[freq >= 0]) / L)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude")
plt.title("Frequency Spectrum (Real Part)")

# change image size and insert gap between subplots
plt.gcf().set_size_inches(10, 8)
plt.tight_layout(h_pad=3.0)

plt.subplot(2, 1, 2)
plt.plot(freq[freq >= 0], np.abs(FFT_x.imag[freq >= 0]) / L)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude")
plt.title("Frequency Spectrum (Imaginary Part)")
plt.show()
```

```python
FFT_detrend_x = np.fft.fft(detrend_x)
L_detrend = len(detrend_x)
freq_detrend = np.fft.fftfreq(len(detrend_x), dt)

# We devided the amplitude by L to normalize the amplitude of the FFT output. This is because the FFT output is scaled by the length of the input signal, and dividing by L ensures that the amplitude of the frequency components is correctly represented in the frequency domain.

plt.subplot(2, 1, 1)
plt.plot(freq_detrend[freq_detrend >= 0], np.abs(FFT_detrend_x.real[freq_detrend >= 0]) / L_detrend)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude")
plt.title("Frequency Spectrum (Real Part)")

# change image size and insert gap between subplots
plt.gcf().set_size_inches(10, 8)
plt.tight_layout(h_pad=3.0)

plt.subplot(2, 1, 2)
plt.plot(freq_detrend[freq_detrend >= 0], np.abs(FFT_detrend_x.imag[freq_detrend >= 0]) / L_detrend)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude")
plt.title("Frequency Spectrum (Imaginary Part)")
plt.show()
```

```python
dt_signal = 0.01
t = np.arange(0, 100, dt_signal)
sig1 = 0.8 * np.sin(2 * np.pi * 7 * t[0: int(len(t)/2)] - np.pi/3)
sig2 = 0.6 * np.cos(2 * np.pi * 12 * t[int(len(t)/2):] + np.pi/4)
sig = np.concatenate((sig1, sig2))

base_sig = 2 * np.sin(2 * np.pi * 1.0 * t) + 1 * np.sin(2 * np.pi * 4 * t + np.pi / 5) + sig

plt.plot(t, base_sig)
plt.title("Synthetic Signal with Multiple Frequencies")
plt.xlabel("Time (Sec)")
plt.ylabel("Amplitude")
plt.show()
```

```python
# Spectrogram
Nfft_signal = 5/dt_signal
frequencies, times, Sxx = signal.spectrogram(base_sig, fs=1/dt_signal, nperseg=256, nfft=Nfft_signal)

plt.pcolormesh(times, frequencies, 10 * np.log10(Sxx), shading="gouraud")
plt.title("Spectrogram of Synthetic Signal")
plt.xlabel("Time (Sec)")
plt.ylabel("Frequency (Hz)")
plt.colorbar(label="Power/Frequency (dB/Hz)")
plt.show()
```
